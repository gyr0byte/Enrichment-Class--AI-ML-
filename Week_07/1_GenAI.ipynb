```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
```

```python
MODEL_NAME = 'google/flan-t5-base'
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
model = AutoModelForSeq2SeqLM.from_pretrained(
MODEL_NAME,
torch_dtype=torch.float32
)
print(f"[INFO] Model successfully initialized in memory with vocabulary size:{tokenizer.vocab_size} tokens.")
```

```python
def generate_text_response(prompt_text:str) -> str:
    input_ids = tokenizer(prompt_text, return_tensors="pt").input_ids
    output_ids = model.generate(
        input_ids,
        max_new_tokens=256,
        min_new_tokens=50,
        num_beams=8,
        early_stopping=True,
        do_sample=True,
        temperature=0.8,
        top_k=50,
        top_p=0.9,
        repetition_penalty=2.0,
        no_repeat_ngram_size=4
 )
    generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    return generated_text
```

```python
document_content = (
 "Generative AI, particularly large language models (LLMs), represents a fundamental paradigm shift in machine learning."
 "Unlike discriminative architectures that evaluate features to assign decision boundaries, generative models learn "
 "the joint probability distribution across massive datasets (like text, images, or audio) to synthesize entirely "
 "new, semantically cohesive sequences or artifacts. This capability allows them to perform tasks such as creative writing, "
 "code generation, image synthesis, and sophisticated summarization, moving beyond mere classification or prediction. "
 "Their architectural advantage lies in their ability to understand and replicate complex data patterns, making them "
 "incredibly versatile tools for a wide array of applications in various industries, from content creation to scientific research."
)
```

```python
document_prompt = f"summarize: {document_content}"
```

```python
print(f"INPUT PROMPT 1 (Summarization Task):\n {document_prompt}")
summary_result = generate_text_response(document_prompt)
print(f"\nModel Response 1: \n {summary_result}")
```

```python
reasoning_question = (
 "What is the main architectural advantage of hosting open-source LLMs"
 " on local infrastructure compared to using closed-source APIs ?"
)
```

```python
reasoning_prompt = f"question: {reasoning_question}"
print(f"INPUT PROMPT 2 (Technical Q&A Task):\n{reasoning_prompt}")
reasoning_result = generate_text_response(reasoning_prompt)
print(f"\nMODEL RESPONSE 2:\n{reasoning_result}")
```
